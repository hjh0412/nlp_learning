```python
# 导包
from sklearn.datasets import load_iris
from sklearn.preprocessing import  StandardScaler     # 数据标准化
from sklearn.neighbors import KNeighborsClassifier     # KNN近邻分类
from sklearn.model_selection import train_test_split    # 数据集划分
from sklearn.model_selection import GridSearchCV    # 网格搜索
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
# 1.导入数据
data = load_iris()
# 2.划分训练集
x_train, x_test, y_train, y_test = train_test_split(data.data, data.target, test_size=0.2, random_state=42)
# 3.特征处理
transform = StandardScaler()
x_train = transform.fit_transform(x_train)
x_test = transform.transform(x_test)
```

```python
# 模型训练 
model =  KNeighborsClassifier()
# 进行网格搜索的超参数 最好定义一个变量 因为后续可能有多个超参数 写在函数里不美观
param_dict = {"n_neighbors":[1,2,3,4,5,6,7]}
# 对模型进行网格搜索和交叉验证
model = GridSearchCV(estimator=model,param_grid=param_dict,cv=5)
# 模型训练
model.fit(x_train,y_train)
```

```python
print(f'交叉验证中最好的结果：{model.best_score_}')
print(f'最好的参数模型：{model.best_estimator_}')
print(f'每次交叉验证后的验证集准确率和训练集准确率结果：{model.cv_results_}')
print(model.score(x_test, y_test))
```
