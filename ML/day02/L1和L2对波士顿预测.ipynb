```python
"""
L1和L2正则化都是为了解决过拟合的问题的 通过降低某个干扰特征值的权重实现
"""
# 导入相关的包
import numpy as np
from sklearn.linear_model import Lasso,Ridge
from sklearn.linear_model import LinearRegression
from sklearn.linear_model import SGDRegressor
from sklearn.datasets import fetch_openml
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_absolute_error
```

```python
# 数据导入
data,target = fetch_openml(name='boston',version=1,return_X_y=True,data_home='../data')
```

```python
# 划分训练集测试集
x_train, x_test, y_train, y_test = train_test_split(data, target, test_size=0.2, random_state=42)
```

```python
# 将数据标准化 否则后续进行L1正则的时候会报错
transform = StandardScaler()
x_train = transform.fit_transform(x_train)
x_test = transform.transform(x_test)
```

```python
# L1 正则化的实现
modul_L1 = Lasso(alpha=0.6)
modul_L1.fit(x_train, y_train)
y_pred_L1 = modul_L1.predict(x_test)
print(f"L1正则化模型在测试集上的预测结果为：{y_pred_L1}")  
print(f"L1正则化的权重为：{modul_L1.coef_}")
```

```python
# L2 正则化预测
# 创建模型
modul_L2 = Ridge(alpha=100)
# 训练模型
modul_L2.fit(x_train,y_train)
# 模型评估/模型预测
y_pred_L2= modul_L2.predict(x_test)
print(f"L2正则化模型在测试集上的预测结果为：{y_pred_L2}")
print(f"L2正则化的权重为：{modul_L2.coef_}")
```
