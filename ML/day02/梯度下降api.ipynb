```python
"""
案例： 线性回归-波士顿房价预测-梯度下降法
线性回归：   用回归方程 对 一个或多个自变量（特征值） 和 因变量（标签）的关系建模， 用建好的模型预测新数据
       公式：  y = w1 * x1 + w2 * x2 + w3 * x3 + .... +b =. w的转置 * x +b
线性回归的分类：
     一元线性回归：  一个特征值
     公式：   y = w * x + b  w:  权重，系数， coef,   b:  偏置， 截距， intercept
     多元线性回归： 多个特征值
     公式：  y = w1 * x1 + w2 * x2 + w3 * x3 + .... +b =. w的转置 * x +b

损失函数：   衡量预测值和真实值之间的效果
       误差= 预测值 -  真实值
        分类：  MAE ：  |误差|求和/ 样本数量
               MSE ： 误差的平方求和/ 样本数量
               最小二乘法：  误差的平方求和

模型评估：
       MAE:
       MSE:
       RMSE: MSE开平方

建模流程：
1. 准备数据
2. 数据预处理
3. 特征工程
4. 模型训练 进行梯度下降
5. 模型评估
"""
# 导包
from sklearn.datasets import fetch_openml    # 导入数据集 获取数据用的
from sklearn.preprocessing import StandardScaler  # 数据标准化
from sklearn.linear_model import SGDRegressor  # 导入梯度下降模型
from sklearn.linear_model import LinearRegression  # 导入线性回归模型
from sklearn.metrics import mean_absolute_error, mean_squared_error, root_mean_squared_error  # 导入模型评估指标
from sklearn.model_selection import train_test_split  # 导入数据集划分函数
import numpy as np  # 导入numpy库
```

```python
# 1.数据准备
data,target = fetch_openml(name='boston',version=1,return_X_y=True,data_home='../data')
print(data[0:5],target[0:5])
```

```python
# 2.数据预处理
# 3.特征工程
# 数据切割
x_train, x_test, y_train, y_test = train_test_split(data,target,test_size=0.2,random_state=42)
# 数据标准化
transform = StandardScaler()
# 标准化训练集
x_train = transform.fit_transform(x_train)
# 标准化测试集
x_test = transform.transform(x_test)
```

```python
# 4.模型训练
# 创建一个梯度下降模型 
# 几个参数：损失函数：平方误差 学习率：常数 学习率：0.04 迭代次数：100
model = SGDRegressor(loss='squared_error', learning_rate='constant', eta0=0.04, max_iter=100)
model.fit(x_train, y_train)
# 查看模型参数
print(model.coef_)            #斜率
print(model.intercept_)       #截距
```

```python
# 5.模型评估
y_pred = model.predict(x_test)
```

```python
error = mean_squared_error(y_test, y_pred)
print(f"均方误差: {error:.2f}")
```
