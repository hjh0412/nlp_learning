```python
# 过拟合和欠拟合演示
# 导包
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
```

```python
# 通过numpy生成一个训练集 随机生成一个范围在-3到3之间的100个数
x = np.random.uniform(-3,3,size=100)
# 目标噪声
y = 0.5 * x ** 2 + x + 2 + np.random.normal(size=100)
#  将x转为了二维数组 不然无法进行模型训练
x_train = x.reshape(-1,1)
```

```python
# 创建模型对象
modul_l = LinearRegression()
# 训练模型
modul_l.fit(x_train,y)
# 模型预测
y_pred = modul_l.predict(x_train)
```

```python
# 绘制图 查看欠拟合的状态
# 描绘散点图的函数
plt.scatter(x, y)
# 绘制预测值的函数 即横纵坐标之间的关系
plt.plot(x_train,y_pred,color='red')
plt.show()
```

```python
# 模拟过拟合的状态
# 要模拟的对象还y函数
# 通过水平上增加幂的方式让模型变得更加复杂模拟过拟合情况 
x_train1 = np.hstack([x_train,x_train**2,x_train**3,x_train**4,x_train**5,x_train**6,x_train**7,x_train**8,x_train**9,x_train**10])
```

```python
# 创建模型
model= LinearRegression()
# 训练模型
model.fit(x_train1,y)
# 预测模型
y_pred1 = model.predict(x_train1)
```

```python
# 绘制图 查看过拟合的状态
# 描绘散点图的函数
plt.scatter(x, y)
# 绘制预测值的函数 即横纵坐标之间的关系
plt.plot(np.sort(x), y_pred1[np.argsort(x)], color='red')
plt.show()
```
