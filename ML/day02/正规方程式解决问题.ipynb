```python
"""
案例： 线性回归-波士顿房价预测-正规方程法
线性回归：   用回归方程 对 一个或多个自变量（特征值） 和 因变量（标签）的关系建模， 用建好的模型预测新数据
       公式：  y = w1 * x1 + w2 * x2 + w3 * x3 + .... +b =. w的转置 * x +b
线性回归的分类：
     一元线性回归：  一个特征值
     公式：   y = w * x + b  w:  权重，系数， coef,   b:  偏置， 截距， intercept
     多元线性回归： 多个特征值
     公式：  y = w1 * x1 + w2 * x2 + w3 * x3 + .... +b =. w的转置 * x +b

损失函数：   衡量预测值和真实值之间的效果
       误差= 预测值 -  真实值
        分类：  MAE ：  |误差|求和/ 样本数量
               MSE ： 误差的平方求和/ 样本数量
               最小二乘法：  误差的平方求和

模型评估：
       MAE:
       MSE:
       RMSE: MSE开平方

建模流程：
1. 准备数据
2. 数据预处理
3. 特征工程
4. 模型训练
5. 模型评估
"""
# 导包
from sklearn.datasets import fetch_openml    # 导入数据集 获取数据用的
from sklearn.preprocessing import StandardScaler  # 数据标准化
from sklearn.linear_model import LinearRegression  # 导入线性回归模型
from sklearn.metrics import mean_absolute_error, mean_squared_error, root_mean_squared_error  # 导入模型评估指标
from sklearn.model_selection import train_test_split  # 导入数据集划分函数
import numpy as np  # 导入numpy库
```

```python
# 1.准备数据
data,target = fetch_openml(name='boston', version=1, return_X_y=True, data_home='../data')  # 获取波士顿房价数据集
print(data[0:5],target[0:5])  # 查看前5条数据
```

```python
# 使用train_test_split函数划分训练集和测试集，test_size=0.2表示测试集占20%，random_state=42表示随机种子，保证每次划分结果一致
x_train, x_test, y_train, y_test = train_test_split(data, target, test_size=0.2, random_state=42)
```

```python
# 2.数据预处理 （这里不需要）
# 3.特征工程
transform = StandardScaler()  # 创建标准化对象
x_train = transform.fit_transform(x_train)  # 对训练集进行标准化处理
x_test = transform.transform(x_test)  # 对测试集进行标准化处理
```

```python
# 4.模型训练
model = LinearRegression()  # 创建线性回归模型对象
model.fit(x_train, y_train)  # 使用训练集数据进行模型训练
print("模型的权重系数为：", model.coef_)  # 输出模型的权重系数
print("模型的偏置为：", model.intercept_)  # 输出模型的截距理解为
```

```python
# 5.模型评估
y_predict = model.predict(x_test)  # 使用测试集数据进行预测
print("预测值为：", y_predict)  # 输出预测值
```

```python
# 计算平均绝对误差
error_mae = mean_absolute_error(y_test, y_predict)  # 计算MAE
print("MAE为：", error_mae)  # 输出MAE
```

```python
# 计算平均误差平方和（MSE）和均方根误差（RMSE）
error_mse = mean_squared_error(y_test, y_predict)  # 计算MSE
print("MSE为：", error_mse)  # 输出MSE
error_rmse = np.sqrt(error_mse)  # 计算RMSE
print("RMSE为：", error_rmse)  # 输出RMSE
```
