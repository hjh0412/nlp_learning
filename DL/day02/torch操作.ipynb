```python
import torch
t1 = torch.randint(1,25,(4,5))
print(f"t1：{t1}")
t2 = torch.randint(1,25,(2,3,4))
print(f"t2：{t2}")
```

```python
"""
案例:
    演示张量的索引操作, 即: 根据需求 获取 张量中的部分数据.

分类:
    简单行列索引
    列表索引
    范围索引
    布尔索引
    多维索引
"""
# 行列索引
print(f"t1的第一行第一列数据：{t1[0,0]}")                        # 返回一个标量
print(f"t1的第一列数据：{t1[:,0]}")
# 列表索引
print(f"t1第一行的第二个数据和第二列的第三个数据：{t1[[0,1],[1,2]]}")
# 下面的两个数据是一个转置矩阵
print(f"t1第一行的二三列和第二行的二三列：{t1[[[0],[1]],[1,2]]}")
print(f"t1第一二列的第一二行数据：{t1[[0,1],[[1],[2]]]}")
# 范围索引
print(f"t1前三列前两行的数据：{t1[:3,:2]}")                         # 左闭右开的
print(f"t1三行后和三列后的数据：{t1[2:,2:]}")
# 布尔索引
print(f"第一行大于20的数据：{t1[0,t1[0]>20]}")
# 千万不能将t1[0]>5放到前面 因为t1[0]>5是个布尔掩码 它的长度是True的长度 会造成形状不匹配（行数必须相同）的情况 会报错！
print(f"第一列大于5的行的数据：{t1[t1[:,0]>5]}")
print(f"第一列大于5的数据：{t1[t1[:,0]>5,0]}")
# 多维索引
print(f"t2 0轴的第一个数据{t2[0,:,:]}")
print(f"t2 1轴的第一个数据{t2[:,0,:]}")
print(f"t2 2轴的第一个数据{t2[:,:,0]}")
```

```python
# 用于进行张量形状练习的张量
t3 = torch.randint(1,20,(3,4))
print(f"t3：{t3}")
print(t3.shape)
print(t3.size())
```

```python
# reshape()函数 转换的时候矩阵的规模要能整除 且转换的新的形状数必须与原来的一样
t3_re1 = t3.reshape(4,3)
t3_re2 = t3.reshape(1,-1)       # 1表示1行，-1表示自动计算列数
print(f"t3_re1的形状{t3_re1.shape}")
print(f"t3_re1:{t3_re1}")
print(f"t3_re2的形状{t3_re2.shape}")
print(f"t3_re2:{t3_re2}")
```

```python
"""
squeeze()和unsqueeze()
1. squeeze()：默认的是删除维度为1的维度
2. unsqueeze()：在指定维度插入维度为1的维度
"""
t3_unsq = t3.unsqueeze(dim=2)
print(t3_unsq.shape)
print(t3_unsq)
t3_sq = t3_unsq.squeeze(dim=2)
print(t3_sq.shape)
print(t3_sq)
```

```python
# 创建用于测试转置数据
t4 = torch.randint(1,10,(2,3,4))
t5 = torch.randint(1,10,(3,4))
print(f"t4的形状{t4.shape}")
print(f"t4:{t4}")
print(f"t5的形状{t5.shape}")
print(f"t5:{t5}")
```

```python
"""
transpose()和permute()的区别
transpose()是转置，permute()是重排序
transpose()是in-place操作 一次只能交换一个维度，permute()是out-of-place（创建新的张量来接收）操作 一次可以交换多个维度
view() 只有张量在内存是连续的时候才能使用 用于张量转换
"""
print(f"t4是否连续：{t4.is_contiguous()}")
print(f"t5是否连续：{t5.is_contiguous()}")
t5_trans = t5.transpose(0,1)
print(f"t5_trans:{t5_trans}")
print(f"t5_trans是否连续：{t5_trans.is_contiguous()}")
print(f"t5.shape:{t5.shape}")
print(f"t5_trans.shape:{t5_trans.shape}")
# permute直接写交换后的顺序就行了
t5_per = t5.permute(1,0)
print(f"t5_per:{t5_per}")
print(f"t5_per是否连续：{t5_per.is_contiguous()}")
print(f"t5.shape:{t5.shape}")
print(f"t5_per.shape:{t5_per.shape}")
# 调用contiguous()函数可以将tensor转换为连续的内存块
# 如果tensor不是连续的，则会拷贝一份连续的tensor
t5_new = t5_per.contiguous()
print(f"t5_new是否连续：{t5_new.is_contiguous()}")
# t5_view1 = t5_trans.view(3,4) 报错
t5_view = t5_new.view(3,4)

```

```python
# 用于拼接操作的数据
t6 = torch.randint(1,10,(2,5))
t7 = torch.randint(10,25,(2,5))
t8 = torch.tensor([[1,2,3,4,5]])
print(f"t6的形状{t6.shape}")
print(f"t6:{t6}")
print(f"t7的形状{t7.shape}")
print(f"t7:{t7}")
print(f"t8的形状{t8.shape}")
print(f"t8:{t8}")
```

```python
"""
张量的拼接操作
torch.cat() 要求维度必须一样 并且选用维度的时候，其他的两个维度要相同 否则不能进行拼接
torch.stack() 要求规模必须一样 拼接之后会升维
"""
# 垂直拼接 相当于增加样本
t6_t7_c0 = torch.cat([t6,t7],dim=0)
# 水平拼接 相当于增加特征
t6_t7_c1 = torch.cat([t6,t7],dim=1)
# 拼接同维的张量
t6_t8_c = torch.cat([t6,t8])
print(f"t6_t7垂直拼接:\n {t6_t7_c0}")
print(f"t6_t7水平拼接:\n {t6_t7_c1}")
print(t6_t8_c)

t6_t7_t0 = torch.stack([t6,t7],dim=0)
t6_t7_t1 = torch.stack([t6,t7],dim=1)
print(f"t6_t7垂直拼接:\n {t6_t7_t0}")
print(f"t6_t7水平拼接:\n {t6_t7_t1}") 
```

```python
# 自动微分模块使用的数据
# 相当于传入了一个梯度
w = torch.tensor(10, requires_grad=True, dtype=torch.float)
print(w)
print(w.grad)
```

```python
"""
案例:
    演示梯度值的计算过程.

回顾:
    梯度 = 求导, 求微分, 即: 函数增加最快的方向.
    梯度下降法更新权重:
        W新 = W旧 - 学习率 * 梯度
        W新 = W旧 - lr * grad

y = w * x + b
权重w
损失函数：
y = 2 * w ** 2
梯度： 4 * w
"""
# 定义损失函数
y = 4 * w ** 2
print(f"{y.grad_fn}")
print(f"{y.sum()}")                                         # w的值是10 带入计算得到400 正向传播
y.sum().backward()                                          # 反向传播
print(f"{w.grad}")                                          # 反向传播更新梯度 对y求导带入得到
# 更新权重
w.data = w.data - 0.01 * w.grad
print(f"{w}")
```

```python
# 最佳梯度下降法使用的数据
w1 = torch.tensor(10, requires_grad=True, dtype=torch.float)
print(w1)
print(w1.grad)
```

```python
"""
案例:
    梯度下降法 求最优解.      这个代码目前理解即可, 后续还会经常写, 看到多了就会了.

梯度:
    求导, 求微分, 即: 上山最快的方向.
梯度下降法 更新权重:
    W新 = W旧 - 学习率 * 梯度
        = w - lr * grad
细节:
    因为梯度默认会累加, 所以需要我们手动 清空梯度, 即: x.grad.zero_()

需求:
    求 y = w² + 20 的极小值点, 并打印y是最小值时 w的值(梯度)

计算思路:
    1. 定义点 w=10
    2. 定义函数 y = w² + 20
    3. 利用梯度下降法 循环迭代100次, 求最优解.
        3.1 正向计算(前向传播)
        3.2 梯度清零.
        3.3 反向传播
        3.4 梯度更新.
"""
# 确定损失函数
# y1 = 2 * w1 **2 + 40
# print(f"{y1.grad_fn}")                       
for i in range(100):
    # 损失函数
    y1 = 2 * w1 **2 + 40
    # 梯度清零 初始的时候梯度为零
    if w1.grad is not None:
        w1.grad.zero_()
    # 反向传播
    y1.sum().backward()
    # 更新权重
    w1.data = w1.data - 0.01 * w1.grad
    print(f"第{i+1}次:原来的梯度是： {w1.data}更新后的梯度是： {w1.grad}")
print(f"最后的梯度{w1.grad}")
print(w1)
```

```python
# 自动微分的注意点
x = torch.tensor([1,2], requires_grad=True, dtype=torch.float32)
# 自动微分的向量是不能直接转换为Numpy数组的
# print(x.numpy()) Can't call numpy() on Tensor that requires grad. Use tensor.detach().numpy() instead.
print(x.detach().numpy())
```

```python
"""
案例:
    演示自动微分模块的应用 -> 大家理解即可, 主要是看看做了那些事儿, 后续到ANN(人工神经网络)中, 代码可能会做微调.

y = x * w +b
# 1. 定义输入张量x, 可以理解为: 神经网络某一层的输入数据.
# 2. 定义目标值张量y, 目标值代表了我们期望模型输出的结果.
# 3. 初始化可学习的权重 w 和 偏置 b
# 4. 计算神经网络输入值 z
# 5. 定义损失函数并计算损失.
# 6. 进行自动微分, 根据链式法则计算上述 w和b的梯度.
# 7. 打印w 和 b的梯度, 用于后续更新: w 和 b
"""
from torch.nn import MSELoss
# 输入值
x_m = torch.ones(2,4)
print(x_m)
# 目标值
y_m = torch.zeros(2,8)
print(y_m)
# 设置权重和偏置的初始值
w_m = torch.randn(4,8, requires_grad=True, dtype=torch.float)        # w_m的取值得根据需求设定 这里要进行矩阵乘法
b_m = torch.randn(2,8, requires_grad=True, dtype=torch.float)        # b_m的规模需要根目标值的相同
# 计算神经网络的预测值
z_m = torch.matmul(x_m, w_m) + b_m                       # 这里是进行矩阵乘法
print(z_m.grad_fn)
# 计算损失
loss_fn = MSELoss()
loss = loss_fn(z_m, y_m)
# 反向传播 算梯度
loss.backward()
# 更新权重和偏置 PyTorch会自动计算梯度 使用的是封装的内置方法
print(f"{w_m.grad}")
print(f"{b_m.grad}")
```
