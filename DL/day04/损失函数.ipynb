```python
"""
深度学习的损失函数
损失函数介绍:
    概述/作用:s
        就是用来衡量模型效果好坏的, 即: 衡量预测值 和 真实值之间的关系.
一、回归问题
回归任务损失函数:
1.MAE 绝对误差距离 L1 Loss
    由于其稀疏性 常常作为正则与其他的loss函数使用
2.MSE 平方误差距离 L2 Loss
    由于其平滑性 常常作为正则与其他的loss函数使用
    预测值和目标值相差较大时 会梯度爆炸
3.smooth L1 Loss
    介于MAE和MSE之间 在MAE和MSE之间 平滑
    在|x|的范围之外采取 L1损失函数 避免了梯度爆炸

二、多分类问题
信息熵
1.多分类任务的损失函数:多分类交叉熵损失函数. 使用softmax激活函数
目的:
    最小化 正确类别所对应的预测概率 的对数 的负值.
公式:
    L = -Σ(真实概率 * log(softmax(加权求和结果)))

2.二分类任务的损失函数:二分类交叉熵损失函数. 使用了sigmoid激活函数
目的:
    最小化 正确类别所对应的预测概率 的对数 的负值.
公式:
    L = -(真实概率 * log(sigmoid(加权求和结果)) + (1 - 真实概率) * log(1 - sigmoid(加权求和结果)))
"""
```

```python
# 导包 多分类任务计算函数
import torch
from torch.nn import CrossEntropyLoss
import torch.nn as nn
```

```python
# 索引类型必须使用int64位
y_ture = torch.tensor([[1,0,0],[0,0,1]],dtype=torch.float32)
y_pred = torch.rand(2,3,requires_grad=True,dtype=torch.float32)
# 实例化函数
loss = CrossEntropyLoss()
# 计算损失结果
result = loss(y_pred, y_ture).detach().numpy()
print(f"result:{result}, type:{type(result)}")
print(f"y_pred:\n{y_pred}, y_ture:\n{y_ture}")
```

```python
# 二分类问题的计算 都是一维的张量进行计算
from torch.nn import BCELoss
y_ture = torch.tensor([0,1,0,1],dtype=torch.float32)
y_pred = torch.rand(4,dtype=torch.float32)
# 创建二分类函数对象
loss =  BCELoss()
# 计算损失
loss_value = loss(y_pred, y_ture).detach().numpy()
print(f"loss_value:{loss_value}")
```

```python
# 回归任务的MAE损失函数
# 1 设置真实值和预测值
y_pred = torch.tensor([1.0, 1.0, 1.9], requires_grad=True)    
y_true = torch.tensor([2.0, 2.0, 2.0], dtype=torch.float32)    
# # 2 实例MAE损失对象
loss = nn.L1Loss()
# 3 计算损失值
result = loss(y_pred, y_true)
print(result)
```

```python
# MSE损失函数
# 1 设置真实值和预测值
y_pred = torch.tensor([0.3, 0.2, 0.4], requires_grad=True)    
y_true = torch.tensor([20000000000000000000000.0, 200000000.0, 159500000.0], dtype=torch.float32)    
# # 2 实例MSE损失对象
loss = nn.MSELoss()
# 3 计算损失值
result = loss(y_pred, y_true)
print(result)
```

```python
# Smooth L1损失函数
# 1 设置真实值和预测值
y_pred = torch.tensor([1.0, 1.0, 1.9], requires_grad=True)    
y_true = torch.tensor([20000000000000000000000.0, 200000000.0, 159500000.0], dtype=torch.float32)    
# # 2 实例Smooth L1损失对象
loss = nn.SmoothL1Loss()
# 3 计算损失值
result = loss(y_pred, y_true)
print(result)
```
