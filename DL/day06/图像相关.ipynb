```python
"""
案例:
    演示 CNN入门案例, 绘制: 黑白图.

回顾: 计算机中图像分类表示
    1. 二值图像.
        1个通道(二维矩阵), 像素点: 0 或者 1
    2. 灰度图像.
        1个通道, 像素值: 0-255
    3. 索引图像
        1个通道, 像素值: 0 ~ 255
    4. RGB真彩图
        最常用的, 3通道(3个二维矩阵), RGB这3个二维矩阵, 像素值: 0 ~ 255

需要你掌握的 matplotlib的三个函数:
    imread()        读取图片信息, 并将像素点封装成矩阵, 格式为: (H, W, C), 即: (高, 宽, 通道数)
    imsave()        保存图片
    imshow()        显示图片
"""
```

```python
# 导包
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 绘制全黑的图像
img = np.zeros([200,200,3])
plt.imshow(img)
plt.show()
```

```python
# 绘制全白的图像
img = np.full([200,200,3],255)
plt.imshow(img)
plt.show()
```

```python
# 噪点图像
img = np.random.randint(0,255,(100,100,3))
print(img.shape)
plt.imshow(img)
plt.show()
```

```python
h, w = 100, 100
x = np.linspace(0, 1, w)
y = np.linspace(0, 1, h)
xx, yy = np.meshgrid(x, y)          # 得到每个像素的横纵坐标

img = np.zeros((h, w, 3), dtype=np.uint8)
img[..., 0] = 255                          # 红色通道：从左到右变亮
img[..., 1] = 0                          # 绿色通道：从上到下变亮
img[..., 2] = 255                         # 蓝色通道：固定一半亮度
plt.imsave("../data/purple.jpg",img)
plt.imshow(img)
plt.show()
```

```python
# 加载和保存图像
# 形状：HWC 长宽 通道数
img = plt.imread("../data/img.jpg")
print(f"图形的形状是：{img.shape}")
plt.imshow(img)
plt.axis("off")
plt.show()
```
