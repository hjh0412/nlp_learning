```python
"""
案例:
    演示 卷积神经网络之 卷积层介绍.

卷积神经网络 介绍:
    全称叫: Convolutional neural network, 主要处理 图片, 视频, 音频等数据.
    组成:
        卷积层:    提取图片的局部特征.
        池化层:    降维. 降低运算量.
        全连接层:  处理并输出结果.

卷积层:
    作用:
        提取图片的局部特征.
    名词解释:
        卷积计算: 从原始数据中 获取对应元素 和 卷积核(滤波器)做 点乘, 然后结果相加 -> 存储到 特征图中.
    细节:
        1. 1个卷积核 = 1个神经元 = 1张特征图.
        2. 特征图大小的计算方式:
            N = (W - F + 2P) / S + 1

            N: 特征图大小.
            W: 原图大小
            F: 卷积核大小
            P: 填充
            S: 步长.
"""
```

```python
# 导包
import torch.nn as nn
import torch
import matplotlib.pylab as plt
```

```python
# 卷积层的实现
# 读取图片 读出来的顺序是H W C
img = plt.imread("../data/img.jpg")
print(img.shape)
# 构建卷积层 三个通道对应的是三个颜色大类
# 输入三个通道 输出三个通道 卷积核的size大小 步长 填充的数量默认为零 取1的话二者的形状完全相同
conv = nn.Conv2d(3,3,3,1,1)
# 输入形状和图像处理
img = torch.tensor(img).permute(2,0,1).unsqueeze(0)
print(img.shape)
# 将图像放入卷积层里面 要求放入的顺序是N C H W
cnn_img = conv(img.to(torch.float32))
# 打印特征图的形状
print(cnn_img.shape)
print(cnn_img)
```

```python
"""
案例:
    演示 池化层的 相关API.

涉及到的池化层的API:
    最大池化.   nn.MaxPool2d(kernel_size=2, stride=2, padding=1)
    平均池化.   nn.AvgPool2d(kernel_size=2, stride=2, padding=1)

    参数说明:
        kernel_size: 池化核的大小.
        stride: 步长.
        padding: 在四周加入padding的数量, 默认补: 0
"""
```

```python
# 池化层
maxpool = nn.MaxPool2d(kernel_size=1,stride=1)
avgpool = nn.AvgPool2d(kernel_size=1,stride=1)
maxpool_out = maxpool(cnn_img)
avgpool_out = avgpool(cnn_img)
print(f"maxpool_out:\n{maxpool_out}")
print(f"avgpool_out:\n{avgpool_out}")
```
