```python
"""
回顾: RNN
    全称叫: Recurrent neural network, 循环神经网络, 处理流程: 词嵌入层 + 循环层 + 全连接层(输出层)
其中:
    词嵌入层把词索引映射成词向量，所有词的向量组成一个词向量矩阵。 即: word_to_vector(word_2_vector, word_vec, word2vec)
思路:
    语料(句子,文本) -> 切词,去重, 得到: 词汇表 -> 词向量矩阵.

    
词嵌入的流程：
    初始化词向量：对输入的词
    输入索引：每个词在词表有对应的键 这个键是唯一的
    查找词向量：
    输入到RNN：会根据上下文对词进行输出


词嵌入层：
    nn.Embedding(num_embeddings, embedding_dim)
    num_embeddings: 词嵌入矩阵的行数，即词典大小
    embedding_dim: 词嵌入矩阵的列数，即每个词向量维度
"""
```

```python
import torch.nn as nn
import torch
# 用于分词的包
import jieba
```

```python
text = "北京冬奥的进度条已经过半，不少外国运动员在完成自己的比赛后踏上归途。"
# 文本分词
words =  jieba.lcut(text)
# 分词后去重保留原来的顺序
unique_words:list[str] = []
for word in words:
    if word not in unique_words:
        unique_words.append(word)
print(f"unique_words:{unique_words}")
# 构建嵌入层
embed = nn.Embedding(len(unique_words), 6)
# 嵌入层的权重矩阵
print(embed.weight.shape)
print(embed.weight.data)
# 词语的向量表示
for i,word in enumerate(unique_words):
    word_vec = embed(torch.tensor(i))
    print(f"{word}:{word_vec}")
```

```python
# 生成rnn网络
rnn = nn.RNN(input_size=128, hidden_size=256)
# 生成输入数据
# x是输入数据 参数 句子的长度：包含多少个单词 批次：有多少个句子 单词的维度：128  
x = torch.randn(10, 3, 128)
# h0是初始化隐藏状态 参数：隐藏层的层数 批次大小 隐藏层的维度
h0 = torch.zeros(1, 3, 256)
# 实例化rnn网络 得到预测结果 二者的形状参数与输入的数据和初始化的隐藏状态相同
result,hn = rnn(x, h0)
print(result)
print(result.shape)
print(hn.shape)
```
