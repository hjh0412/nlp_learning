```python
"""
案例:
    演示jieba分词的详细介绍.

分词解释:
    把句子 -> 切割成 词级别, 词是语言语义理解的最基本的单位.

分词方式:
    1. jieba 分词器.
        支持多种分词模式(精确模式, 全模式, 搜索引擎模式), 中文繁体分词, 用户自定义词典.
    2. SnowNLP
        基于概率算法的中文自然语言处理工具.
    3. pyltp
        哈工大的, 适用于: 复杂的自然语言处理任务.
    4. THULAC
        清华的...
    5. ik分词器
        ElasticSearch 搜索引擎用的多.
    上述的分词模式, 其实就是 分词的粒度不同, 例如: "软件工程" -> 软件工程(粗粒度),  软件,工程(细粒度)

# 1. 定义待分词的文本内容.
    sentence = "传智教育是一家上市公司, 旗下有黑马程序员品牌. 我是在黑马这里学习人工智能"

#jieba精确分词模式， 适用于： 文本分析， 自然语言处理
#jieba全模式分词模式， 适用于： 关键词提取， 不需要 严格分词的准确性的场景
#jieba搜索引擎模式分词, 适用于: 搜索引擎, 文本匹配.

解释: 搜索引擎分词模式, 在精确模式分词的基础上, 对长词进行再次切分, 提高召回率.
例如:
    场景1: 用户录入 "程序员"
        精确模式:       只能匹配出包含"程序员"的文档.
        搜索引擎模式:    不仅能匹配'程序员'的文档, 还能匹配'程序', '员'的文档, 提高召回.
    场景2: 用户搜索 '黑马程序'
        精确模式:       无法匹配, 分词为: ('黑马', '程序员')
        搜索引擎模式:    能匹配, 分词为: ('黑马', '程序', '员')
    场景3: 实际应用场景(电商搜索), 商品标题为: "苹果手机保护套", 用户搜索"苹果套"
        精确模式:       无法匹配, 分词为: '苹果', '手机', '保护套'
        搜索引擎模式:    能匹配, 分词为: '苹果', '手机', '保护', '套'

jieba -> 繁体分词, 适用于: 中国香港, 台湾, 澳门...

jieba -> 自定义词典分词, 适用于: 稍微生僻点的词 或者 特殊要求的词组, 一般不会太多.

添加自定义词典后, jieba能够准确识别词典中出现的词汇, 提升整体的识别准确率.
词典格式:
    每一行分3部分: 词语, 词频(可省略), 词性(可省略), 用空格隔开, 顺序不可颠倒.

"""
```

```python
import jieba
sentence = "K124次列车由十堰开往上海松江，由武汉铁路局襄阳客运段担当，车底为配属于武局武段的25G客车，本务是配属于武局襄段的HXD3C"
```

```python
# 全分词模式 将句子精确的切开 适合文本分析 大模型自动分割
words = jieba.cut(sentence,True)
# 方式一、next取出文本
# print(next(words))
# 方法二、通过list
print(list(words))
```

```python
# 全模式分词模式 适用于关键词提取 不需要严格分词 易有歧义
words =  jieba.lcut(sentence,False)
print(list(words))
```

```python
# 搜索引擎模式 在精确模式的基础上再进行分词 适合于浏览器的搜素引擎
words1 = jieba.cut_for_search(sentence)
words2 = jieba.lcut_for_search(sentence)
print(list(words1))
print(list(words2))
```

```python
# 处理繁体字
sentence = "煩惱即是菩提，我暫且不提"
words = jieba.lcut(sentence,True)
print(list(words))
```

```python
# 用户自定义词典
sentence =  '传智教育是一家上市公司，旗下有黑马程序员品牌。我是在黑马这里学习人工智能'
words = jieba.cut(sentence,True)
print(list(words))
# 使用自定义词典后
jieba.load_userdict("userdict.txt")
words1 = jieba.cut(sentence,True)
print(list(words1))
```

```python
# 词性标注
import jieba.posseg as pseg
# sentence = "国庆节快乐"
data = pseg.lcut(sentence)
print(list(data))
```

```python
# one-hot编码
from tensorflow.keras.preprocessing.text import Tokenizer
import jieba
import joblib
```

```python
# 准备语料vocabs
vocabs = {"周杰伦", "陈奕迅", "王力宏", "李宗盛", "吴亦凡", "鹿晗"}
# 实例化映射器
mytokenizer = Tokenizer()
mytokenizer.fit_on_texts(vocabs)
# 查询单词idx，赋值zero_list生成onehot
for vocab in vocabs:
    zero_list = [0]*len(vocabs)
    idx = mytokenizer.word_index[vocab] - 1
    zero_list[idx] = 1
    print(f"{vocab}的编码是{zero_list}")
# 使用joblib工具保存映射器joblib.dump()
mypath = './mytokenizer'
joblib.dump(mytokenizer,mypath)
# 
print(mytokenizer.word_index)
print(mytokenizer.index_word)
```
