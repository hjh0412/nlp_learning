```python
import fasttext
```

```python
# 数据的路径 可能无法识别中文路径
data_dir = "./data/bj30ad"
```

```python
# 模型训练
model = fasttext.train_unsupervised(data_dir, thread=10, lr=0.05, model="skipgram", dim=10, epoch=5)
```

```python
# 保存模型
model.save_model("./model/fasttext_model.bin")
```

```python
# 模型推理
# 查看词向量
vocabulary = model.get_word_vector("music")
# 查看与其相近的词是什么 只看排名最靠前的2个
neig_voc = model.get_nearest_neighbors("music",2)
```

```python
print(vocabulary)
print(neig_voc)
print(type(neig_voc))
```
